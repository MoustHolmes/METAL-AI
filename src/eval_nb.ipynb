```python
import torch
import os
import sys
from typing import Any, Dict

# Add the project root to the path to allow importing from src
# Adjust this to point to your project root directory
PROJECT_ROOT = "/Users/moustholmes/Projects/METAL-AI"
sys.path.append(PROJECT_ROOT)

# Function to load a model from a checkpoint
def load_model_from_checkpoint(checkpoint_path):
    """Load a model from a Lightning checkpoint file."""
    print(f"Loading checkpoint from: {checkpoint_path}")
    
    try:
        # First try the simple approach - this often works for newer Lightning versions
        checkpoint = torch.load(checkpoint_path, map_location=torch.device('cpu'))
        
        # Check if this is a Lightning checkpoint or just a state dict
        if 'state_dict' in checkpoint:
            print("Found Lightning checkpoint with state_dict")
            state_dict = checkpoint['state_dict']
        else:
            print("Assuming checkpoint is already a state dict")
            state_dict = checkpoint
            
        print(f"Successfully loaded state dict with {len(state_dict)} parameters")
        return state_dict
        
    except Exception as e:
        print(f"Error loading checkpoint directly: {e}")
        print("Trying alternative approach...")
        
        # If there's an error, we'll try to create the model first
        # This requires importing your model classes from src
        try:
            # Import your model definition
            from src.models.components.simple_CSF_encoder import simple_CSF_encoder
            from src.models.simple_transformer_encoder_model import simple_transformer_encoder_model
            
            # Create the model with the same architecture
            csf_encoder = simple_CSF_encoder(output_size=64)
            model = simple_transformer_encoder_model(
                csf_encoder=csf_encoder,
                d_model=32,
                nhead=8,
                dim_forward=64,
                num_layers=8,
                dropout=0.0,
                output_activation=torch.nn.ReLU(),
                output_size=1
            )
            
            # Now, try to load the checkpoint into this model
            checkpoint = torch.load(checkpoint_path, map_location=torch.device('cpu'))
            
            if 'state_dict' in checkpoint:
                model.load_state_dict(checkpoint['state_dict'])
                print("Successfully loaded checkpoint into model")
                return model.state_dict()
            else:
                model.load_state_dict(checkpoint)
                print("Successfully loaded state dict into model")
                return model.state_dict()
                
        except Exception as e2:
            print(f"Error creating model and loading checkpoint: {e2}")
            return None

# Function to export the state dict to a clean format
def export_model_weights(state_dict, output_path):
    """Export model weights to a clean PyTorch file."""
    if state_dict is None:
        print("No state dict to export")
        return False
    
    # Clean the state dict by removing any 'module.' prefixes (from DataParallel)
    clean_state_dict = {}
    for k, v in state_dict.items():
        name = k
        if name.startswith('module.'):
            name = name[7:]  # remove 'module.' prefix
        clean_state_dict[name] = v
    
    # Save the clean state dict
    try:
        torch.save(clean_state_dict, output_path)
        print(f"Successfully exported weights to: {output_path}")
        print(f"Absolute path: {os.path.abspath(output_path)}")
        return True
    except Exception as e:
        print(f"Error saving weights: {e}")
        return False

# Main function to run in your notebook
def extract_and_save_weights(checkpoint_path, output_path):
    """Load a checkpoint and save just the model weights."""
    # Load the model weights
    state_dict = load_model_from_checkpoint(checkpoint_path)
    
    # Export the weights
    if state_dict is not None:
        success = export_model_weights(state_dict, output_path)
        return success
    return False

# Example usage in your notebook:
extract_and_save_weights("/Users/moustholmes/Projects/METAL-AI/best_models/epoch_2968.ckpt", "clean_model_weights.pt")
```
